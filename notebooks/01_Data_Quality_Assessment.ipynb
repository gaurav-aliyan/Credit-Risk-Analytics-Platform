### Objective
The objective of this notebook is to assess the quality of the loan application dataset before performing feature engineering, exploratory data analysis, and predictive modeling. The assessment includes data structure, missing values, duplicates, data types, class imbalance, and descriptive statistics.

```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": round((df.isnull().sum()/len(df))*100,2),
    "Data Type": df.dtypes
})

missing = missing[missing["Missing Values"]>0]
missing = missing.sort_values("Missing Percentage",ascending=False)

missing.head(20)
```

```python
from pathlib import Path

folders = [
    "../reports/Phase_2",
    "../images",
    "../data/processed"
]

for folder in folders:
    Path(folder).mkdir(parents=True, exist_ok=True)

print("Project folders are ready.")
```

```python
missing.to_excel("../reports/Phase_2/Missing_Value_Report.xlsx", index=True)
```

```python
plt.figure(figsize=(12,8))

plt.barh(
    missing.head(20).index,
    missing.head(20)["Missing Percentage"]
)

plt.xlabel("Missing Percentage")
plt.title("Top 20 Columns with Missing Values")

plt.savefig(
    "../images/missing_values.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
duplicates = df.duplicated().sum()

print(f"Duplicate Rows : {duplicates}")
```

```python
numerical = df.select_dtypes(include=["int64","float64"])

categorical = df.select_dtypes(include=["object"])

print(f"Numerical Features : {numerical.shape[1]}")
print(f"Categorical Features : {categorical.shape[1]}")
```

```python
target = df["TARGET"].value_counts()

target
```

```python
plt.figure(figsize=(6,5))

plt.bar(
    target.index.astype(str),
    target.values
)

plt.title("Loan Default Distribution")
plt.xlabel("TARGET")
plt.ylabel("Count")

plt.savefig(
    "../images/Loan Default Distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

```

### Observation

TARGET = 0 → Customer repaid the loan.

TARGET = 1 → Customer defaulted.

The dataset is imbalanced because non-default cases are significantly higher than default cases.

## Key Findings

- No duplicate records (if your result shows 0).
