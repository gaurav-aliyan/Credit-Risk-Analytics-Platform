```python
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": round(df.isnull().mean() * 100, 2),
    "Data Type": df.dtypes
})

missing = missing[missing["Missing Values"] > 0]
missing = missing.sort_values("Missing Percentage", ascending=False)

missing.head(30)
```

```python
missing.to_excel(
    "../reports/Phase_3/Missing_Value_Treatment_Report.xlsx",
    index=True
)

print("Missing value report saved successfully.")
```

```python
high_missing = missing[missing["Missing Percentage"] > 60]
medium_missing = missing[
    (missing["Missing Percentage"] >= 20) &
    (missing["Missing Percentage"] <= 60)
]
low_missing = missing[missing["Missing Percentage"] < 20]

print("Columns with >60% missing :", len(high_missing))
print("Columns with 20-60% missing :", len(medium_missing))
print("Columns with <20% missing :", len(low_missing))
```

| Missing % | Action                                         |
| --------- | ---------------------------------------------- |
| > 60%     | Drop the column                                |
| 20–60%    | Impute carefully (depends on business meaning) |
| < 20%     | Fill missing values                            |


```python
cols_to_drop = [
    "COMMONAREA_MEDI",
    "COMMONAREA_AVG",
    "COMMONAREA_MODE",
    "NONLIVINGAPARTMENTS_MEDI",
    "NONLIVINGAPARTMENTS_MODE",
    "NONLIVINGAPARTMENTS_AVG",
    "FONDKAPREMONT_MODE",
    "LIVINGAPARTMENTS_MODE",
    "LIVINGAPARTMENTS_MEDI",
    "LIVINGAPARTMENTS_AVG",
    "FLOORSMIN_MODE",
    "FLOORSMIN_MEDI",
    "FLOORSMIN_AVG",
    "YEARS_BUILD_MODE",
    "YEARS_BUILD_MEDI",
    "YEARS_BUILD_AVG"
]

df.drop(columns=cols_to_drop, inplace=True)

print(f"{len(cols_to_drop)} columns dropped.")
print(df.shape)
```

```python
num_cols = df.select_dtypes(include=["int64", "float64"]).columns

for col in num_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].median(), inplace=True)
```

```python
remaining_missing = df.isnull().sum().sum()

print(f"Remaining Missing Values: {remaining_missing}")
```

```python
duplicates = df.duplicated().sum()

print(f"Duplicate Rows: {duplicates}")
```

### Handling Special Values

The `DAYS_EMPLOYED` column contained the placeholder value `365243`, which represents missing or unknown employment duration in the Home Credit dataset.

This value was replaced with `NaN` to prevent it from biasing descriptive statistics and downstream machine learning models.

```python
df["DAYS_EMPLOYED"] = df["DAYS_EMPLOYED"].replace(365243, np.nan)
```

```python
print("Remaining placeholder values:", (df["DAYS_EMPLOYED"] == 365243).sum())
```

```python
df["DAYS_EMPLOYED"].describe()
```

```python
financial_cols = [
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE"
]
```

```python
for col in financial_cols:
    plt.figure(figsize=(8, 4))
    plt.boxplot(df[col].dropna(), vert=False)
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)

    plt.savefig(
    "../images/Boxplots.png",
    dpi=300,
    bbox_inches="tight"
)

    plt.show()
```

```python
df["AGE"] = (-df["DAYS_BIRTH"] / 365).round(1)

df[["DAYS_BIRTH", "AGE"]].head()
```

```python
df["YEARS_EMPLOYED"] = (-df["DAYS_EMPLOYED"] / 365).round(1)

df[["DAYS_EMPLOYED", "YEARS_EMPLOYED"]].head()
```

```python
df["INCOME_PER_FAMILY_MEMBER"] = (
    df["AMT_INCOME_TOTAL"] / df["CNT_FAM_MEMBERS"]
).round(2)
```

```python
df["CREDIT_INCOME_RATIO"] = (
    df["AMT_CREDIT"] / df["AMT_INCOME_TOTAL"]
).round(2)
```

```python
df["ANNUITY_INCOME_RATIO"] = (
    df["AMT_ANNUITY"] / df["AMT_INCOME_TOTAL"]
).round(2)
```

```python
df["GOODS_CREDIT_RATIO"] = (
    df["AMT_GOODS_PRICE"] / df["AMT_CREDIT"]
).round(2)
```

```python
new_features = [
    "AGE",
    "YEARS_EMPLOYED",
    "INCOME_PER_FAMILY_MEMBER",
    "CREDIT_INCOME_RATIO",
    "ANNUITY_INCOME_RATIO",
    "GOODS_CREDIT_RATIO"
]

df[new_features].head()
```
