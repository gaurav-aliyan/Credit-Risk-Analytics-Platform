# Phase 4 — EDA Conclusions

## Key Findings

### 1. Default Rate

The dataset contains 307,511 loan applications, with an overall default rate of 8.07%. The target variable is therefore imbalanced, with non-default applications representing 91.93% of observations.

### 2. Age and Default Risk

Default rates showed a clear decreasing pattern across age groups. The observed default rate was highest among applicants aged 18–25 at 12.27% and lowest among applicants aged 65+ at 3.65%.

This indicates that younger applicants have a higher observed default rate in the dataset.

### 3. External Credit Scores

External credit-score variables showed the strongest negative relationships with default.

EXT_SOURCE_AVG had a correlation of -0.2208 with TARGET, followed by EXT_SOURCE_2 (-0.1603) and EXT_SOURCE_3 (-0.1559).

Defaulting applicants also had a substantially lower average external score than non-default applicants:

- Non-default: 0.5186
- Default: 0.4309

### 4. Employment Duration

YEARS_EMPLOYED showed a negative correlation of -0.0750 with TARGET, indicating that longer employment duration is associated with lower observed default rates.

### 5. Financial Characteristics

The analysis of financial ratios showed differences between default and non-default applicants. The largest difference among the engineered financial features was observed in EXT_SOURCE_AVG, followed by CREDIT_PER_FAMILY_MEMBER and CREDIT_TERM.

### 6. Categorical Characteristics

Default rates varied across income type, education level, housing type, and gender categories. For example, lower-secondary education had a higher observed default rate than higher education, while applicants living in rented apartments showed a higher default rate than applicants in houses or apartments.

These findings represent associations within the dataset and should not be interpreted as causal relationships.

## Business Implications

The EDA indicates that external credit scores, applicant age, employment duration, and financial characteristics provide useful signals for differentiating applicants with different observed default rates.

These findings will guide the feature engineering and machine learning phases of the project.
