```python
feature_df = df.copy()
```

```python
binary_cols = [
    "CODE_GENDER",
    "FLAG_OWN_CAR",
    "FLAG_OWN_REALTY",
    "EMERGENCYSTATE_MODE"
]

feature_df[binary_cols].head()
```

```python
feature_df["CODE_GENDER"] = feature_df["CODE_GENDER"].map({
    "M": 1,
    "F": 0
})
```

```python
feature_df["FLAG_OWN_CAR"] = feature_df["FLAG_OWN_CAR"].map({
    "Y": 1,
    "N": 0
})
```

```python
feature_df["FLAG_OWN_REALTY"] = feature_df["FLAG_OWN_REALTY"].map({
    "Y": 1,
    "N": 0
})
```

```python
feature_df["EMERGENCYSTATE_MODE"] = feature_df["EMERGENCYSTATE_MODE"].map({
    "Yes": 1,
    "No": 0
})
```

```python
feature_df[binary_cols].head()
```

```python
feature_df.select_dtypes(include="object").columns.tolist()
```

```python
cat_cols = feature_df.select_dtypes(include="object").columns.tolist()

print(f"Number of categorical columns: {len(cat_cols)}")
print(cat_cols)
```

```python
feature_df = pd.get_dummies(
    feature_df,
    columns=cat_cols,
    drop_first=True,
    dtype="int8"
)
```

```python
print(feature_df.shape)
```

```python
feature_df.select_dtypes(include="object").columns
```

```python
feature_df.dtypes.value_counts()
```

```python
feature_df["LOAN_TO_INCOME_RATIO"] = (
    feature_df["AMT_CREDIT"] /
    feature_df["AMT_INCOME_TOTAL"]
)
```

```python
feature_df["ANNUITY_BURDEN"] = (
    feature_df["AMT_ANNUITY"] /
    feature_df["AMT_INCOME_TOTAL"]
)
```

```python
feature_df["CREDIT_PER_CHILD"] = (
    feature_df["AMT_CREDIT"] /
    (feature_df["CNT_CHILDREN"] + 1)
)
```

```python
feature_df["INCOME_PER_CREDIT"] = (
    feature_df["AMT_INCOME_TOTAL"] /
    feature_df["AMT_CREDIT"]
)
```

```python
feature_df["CREDIT_GOODS_DIFF"] = (
    feature_df["AMT_CREDIT"] -
    feature_df["AMT_GOODS_PRICE"]
)
```

```python
feature_df["CREDIT_GOODS_RATIO"] = (
    feature_df["AMT_CREDIT"] /
    feature_df["AMT_GOODS_PRICE"]
)
```

```python
feature_df["INCOME_STABILITY"] = (
    feature_df["AMT_INCOME_TOTAL"] /
    (feature_df["YEARS_EMPLOYED"] + 1)
)
```

```python
feature_df["FAMILY_BURDEN"] = (
    feature_df["CNT_CHILDREN"] /
    feature_df["CNT_FAM_MEMBERS"]
)
```

```python
feature_df["EMPLOYMENT_PERCENT"] = (
    feature_df["YEARS_EMPLOYED"] /
    feature_df["AGE"]
)
```

```python
feature_df["CREDIT_DENSITY"] = (
    feature_df["AMT_CREDIT"] /
    feature_df["AGE"]
)
```

```python
new_features = [
    "LOAN_TO_INCOME_RATIO",
    "ANNUITY_BURDEN",
    "CREDIT_PER_CHILD",
    "INCOME_PER_CREDIT",
    "CREDIT_GOODS_DIFF",
    "CREDIT_GOODS_RATIO",
    "INCOME_STABILITY",
    "FAMILY_BURDEN",
    "EMPLOYMENT_PERCENT",
    "CREDIT_DENSITY"
]

feature_df[new_features].describe().T
```

```python
feature_df["INCOME_STABILITY"] = feature_df["INCOME_STABILITY"].fillna(0)
feature_df["EMPLOYMENT_PERCENT"] = feature_df["EMPLOYMENT_PERCENT"].fillna(0)
```

```python
print("Current Shape:", feature_df.shape)
```

```python
corr_matrix = feature_df.corr().abs()

upper = corr_matrix.where(
    np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
)

high_corr = [
    column for column in upper.columns
    if any(upper[column] > 0.95)
]

print("Highly Correlated Features:", len(high_corr))
high_corr
```

```python
from sklearn.feature_selection import VarianceThreshold

selector = VarianceThreshold(threshold=0.01)
selector.fit(feature_df.drop(columns="TARGET"))

low_variance = feature_df.drop(columns="TARGET").columns[
    ~selector.get_support()
]

print("Low Variance Features:", len(low_variance))
list(low_variance)
```

```python
constant_cols = [
    col for col in feature_df.columns
    if feature_df[col].nunique() == 1
]

print("Constant Columns:", len(constant_cols))
constant_cols
```

```python
duplicate_cols = []

cols = feature_df.columns

for i in range(len(cols)):
    for j in range(i + 1, len(cols)):
        if feature_df[cols[i]].equals(feature_df[cols[j]]):
            duplicate_cols.append(cols[j])

print("Duplicate Columns:", len(duplicate_cols))
duplicate_cols
```

```python
print("Original Features :", feature_df.shape[1])
print("Highly Correlated :", len(high_corr))
print("Low Variance      :", len(low_variance))
print("Constant Columns  :", len(constant_cols))
print("Duplicate Columns :", len(duplicate_cols))
```

```python
feature_df = feature_df.drop(columns=drop_corr)

print("New Shape:", feature_df.shape)
```
